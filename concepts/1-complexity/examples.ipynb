## Example 1: Pair Search
Let's analyze the following function:

```python
def findPair(a, n, z):
    for i in range(n):

        for j in range(n):

            if (i != j and a[i] + a[j] == z):
                return True

    return False

a = [1, -2, 1, 0, 5]
z = 0
n = len(a)

if findPair(a, n, z):
    print("True")
else:
    print("False")
```

### Execution Analysis
1. **Outer Loop:** `for i in range(n)` iterates `n` times.
2. **Inner Loop:** For each iteration of `i`, `for j in range(n)` also iterates `n` times.
   - This results in **n × n = n^2** iterations in total.
3. **Condition Check:** `if (i != j and a[i] + a[j] == z)` is performed for each pair `(i, j)`, taking O(1) time.
4. **Early Exit:** If a valid pair is found, the function exits early with `return True`.

### Time Complexity
- Worst-case scenario: No pair adds up to `z`. The function evaluates all `n^2` iterations. 
- Time complexity: **O(n^2)**.

### Space Complexity
- No additional data structures are used.
- Space complexity: **O(1)** (constant space).

## Optimized Approach
To improve efficiency:
1. Use a hash table (dictionary) to store visited elements.
2. Check if the complement (`z - a[i]`) exists in the hash table.

```python
def findPairOptimized(a, z):
    seen = set()
    for num in a:
        if z - num in seen:
            return True
        seen.add(num)
    return False

a = [1, -2, 1, 0, 5]
z = 0

if findPairOptimized(a, z):
    print("True")
else:
    print("False")
```

1. **Time Complexity:**
   - Single loop: **O(n)**.
2. **Space Complexity:**
   - Hash table: **O(n)**.

## Example 2: Matrix Multiplication
Matrix multiplication involves multiplying two matrices of size `n × n`.

```python
def matrixMultiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
result = matrixMultiplication(A, B)
print(result)
```

#### Execution Analysis:
1. **Outer Loop:** `for i in range(n)` iterates `n` times.
2. **Middle Loop:** `for j in range(n)` iterates `n` times for each `i`.
3. **Inner Loop:** `for k in range(n)` iterates `n` times for each pair `(i, j)`.
   - This results in **n × n × n = n^3** iterations in total.
4. **Operation:** `C[i][j] += A[i][k] * B[k][j]` is O(1).

#### Time Complexity:
- **O(n^3)**.

#### Space Complexity:
- Space used for the result matrix `C`: **O(n^2)**.

## Example 3: Fibonacci Sequence (Recursive)
Recursive Fibonacci calculation has exponential complexity due to repeated calculations.

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

n = 5
print(fibonacci(n))
```

#### Execution Analysis:
1. **Recursive Calls:** Each call to `fibonacci(n)` makes two recursive calls to `fibonacci(n - 1)` and `fibonacci(n - 2)`.
2. **Tree Depth:** The depth of the recursion tree is `n`.
3. **Total Calls:** The total number of calls grows exponentially, approximately **2^n**.

#### Time Complexity:
- **O(2^n)**.

#### Space Complexity:
- Stack depth is `n`.
- **O(n)**.

## Optimization:
Use memoization or dynamic programming to store previously computed values.

```python
def fibonacciDP(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fibonacciDP(n - 1, memo) + fibonacciDP(n - 2, memo)
    return memo[n]

n = 5
print(fibonacciDP(n))
```

#### Time Complexity (Optimized):
- **O(n)**.

#### Space Complexity (Optimized):
- **O(n)**.

### Example 4: Traveling Salesperson Problem (TSP)
This problem involves finding the shortest path visiting all cities exactly once and returning to the starting point. The brute-force solution tries all permutations of cities.


```python
from itertools import permutations

def tspBruteForce(graph):
    n = len(graph)
    min_cost = float('inf')
    for perm in permutations(range(n)):
        cost = 0
        for i in range(n - 1):
            cost += graph[perm[i]][perm[i + 1]]
        cost += graph[perm[-1]][perm[0]]
        min_cost = min(min_cost, cost)
    return min_cost

graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

print(tspBruteForce(graph))
```

#### Execution Analysis:
1. **Permutations:** There are `n!` permutations of cities.
2. **Cost Calculation:** For each permutation, the cost is calculated in **O(n)**.

#### Time Complexity:
- **O(n! × n)**.

#### Space Complexity:
- **O(n)** for storing permutations.



## Optimization:

Dynamic programming (Held-Karp algorithm) reduces complexity to **O(n^2 × 2^n)**.

### Optimized TSP with Dynamic Programming

```python
# (1 << n) == 2^n
# visited starts at 0001, then 0011, then 0111, then 1111
from functools import lru_cache

def tspDP(graph):
    n = len(graph)

    @lru_cache(None)
    def visit(city, visited):
        if visited == (1 << n) - 1:  # All cities visited
            return graph[city][0] or float('inf')

        min_cost = float('inf')
        for next_city in range(n):
            if not visited & (1 << next_city):
                cost = graph[city][next_city] + visit(next_city, visited | (1 << next_city))
                min_cost = min(min_cost, cost)
            

        return min_cost

    return visit(0, 1)  # Start at city 0 with only city 0 visited

graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

print(tspDP(graph))
```

#### Execution Analysis:
1. **State Representation:** The algorithm tracks the current city and visited cities as states.
2. **Subproblems:** There are **n × 2^n** subproblems.
3. **Recursive Calls:** Each subproblem computes the minimum cost to visit remaining cities.

#### Time Complexity:
- **O(n^2 × 2^n)**.

#### Space Complexity:
- **O(n × 2^n)** for memoization.
