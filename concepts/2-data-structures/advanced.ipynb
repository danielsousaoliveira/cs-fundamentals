# Advanced Data Structures

Advanced data structures offer efficient solutions for complex problems. These structures extend beyond basic arrays, linked lists, and binary trees, enabling advanced functionalities for specific use cases.

## 1. **Tries (Prefix Trees)**
Tries are tree-like data structures used to store strings efficiently, especially useful for prefix-based queries.

### Key Operations:
1. **Insert:** Add a word to the trie.
2. **Search:** Check if a word or prefix exists in the trie.
3. **Delete:** Remove a word from the trie.

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end_of_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_end_of_word = True

    def search(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]
        return node.is_end_of_word

trie = Trie()
trie.insert("apple")
print(trie.search("apple"))
print(trie.search("app"))
```

### Complexity:
- **Insert/Search/Delete:** O(m), where m is the length of the word.
- **Space Complexity:** O(n * m), where n is the number of words and m is the average word length.

## 2. **Segment Trees**
Segment trees efficiently solve range queries and updates, commonly used for problems involving sums, minimums, or maximums over subarrays.

### Key Operations:
1. **Build:** Construct the tree from an array.
2. **Query:** Find the range result (e.g., sum, min).
3. **Update:** Modify an element in the array and update the tree.

```python
class SegmentTree:
    def __init__(self, data):
        self.n = len(data)
        self.tree = [0] * (2 * self.n)
        self.build(data)

    def build(self, data):
        for i in range(self.n):
            self.tree[self.n + i] = data[i]
        for i in range(self.n - 1, 0, -1):
            self.tree[i] = self.tree[2 * i] + self.tree[2 * i + 1]

    def update(self, idx, value):
        pos = idx + self.n
        self.tree[pos] = value
        while pos > 1:
            pos //= 2
            self.tree[pos] = self.tree[2 * pos] + self.tree[2 * pos + 1]

    def query(self, left, right):
        left += self.n
        right += self.n
        result = 0
        while left < right:
            if left % 2:
                result += self.tree[left]
                left += 1
            if right % 2:
                right -= 1
                result += self.tree[right]
            left //= 2
            right //= 2
        return result

data = [1, 3, 5, 7, 9, 11]
st = SegmentTree(data)
print(st.query(1, 4))  # Sum of elements from index 1 to 3
st.update(2, 6)  # Update index 2 to 6
print(st.query(1, 4))  # Updated sum
```

### Complexity:
- **Build:** O(n)
- **Query/Update:** O(log n)
- **Space Complexity:** O(2n) = O(n)

## 3. **Fenwick Trees (Binary Indexed Trees)**
Fenwick Trees are used for range queries and point updates, offering a simpler alternative to segment trees in some cases.

### Key Operations:
1. **Add:** Update a value.
2. **Sum:** Calculate the prefix sum up to an index.

```python
class FenwickTree:
    def __init__(self, size):
        self.size = size
        self.tree = [0] * (size + 1)

    def add(self, idx, value):
        while idx <= self.size:
            self.tree[idx] += value
            idx += idx & -idx

    def sum(self, idx):
        result = 0
        while idx > 0:
            result += self.tree[idx]
            idx -= idx & -idx
        return result

ft = FenwickTree(5)
ft.add(1, 3)
ft.add(2, 2)
ft.add(3, 1)
print(ft.sum(3))  # Sum of first 3 elements
```

### Complexity:
- **Update/Query:** O(log n)
- **Space Complexity:** O(n)

## 4. **Disjoint Set Union (DSU)**
Disjoint Set Union (or Union-Find) efficiently handles dynamic connectivity queries, commonly used in graph problems like Kruskal’s MST algorithm.

### Key Operations:
1. **Find:** Identify the representative of a set.
2. **Union:** Merge two sets.

```python
class DSU:
    def __init__(self, size):
        self.parent = list(range(size))
        self.rank = [1] * size

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]

    def union(self, x, y):
        root_x = self.find(x)
        root_y = self.find(y)

        if root_x != root_y:
            if self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            elif self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            else:
                self.parent[root_y] = root_x
                self.rank[root_x] += 1

dsu = DSU(5)
dsu.union(0, 1)
dsu.union(1, 2)
print(dsu.find(0) == dsu.find(2)) 
print(dsu.find(0) == dsu.find(3)) 
```

### Complexity:
- **Find/Union:** O(α(n)), where α(n) is the inverse Ackermann function (very close to O(1) in practice).
- **Space Complexity:** O(n)
