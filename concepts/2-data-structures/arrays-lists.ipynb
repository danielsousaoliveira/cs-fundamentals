# Arrays and Lists

## Arrays
An **array** is a collection of elements, all of the same data type, stored in contiguous memory locations. Arrays allow efficient access and manipulation of data.

### Properties:
1. **Fixed Size**: Arrays have a predefined size.
2. **Indexing**: Access elements using their index (0-based).
3. **Homogeneous**: All elements must be of the same type.

### Basic Operations:
- **Accessing Elements**: O(1) time complexity.
- **Inserting/Deleting Elements**: O(n) time complexity (due to shifting).
- **Traversal**: O(n) time complexity.

### Limitations:
- Fixed size: Cannot dynamically grow or shrink. (This does not apply to Python array)
- Limited functionality compared to Python lists. (It can only have homogeneous elements)

```python
### Example in Python:
### Although Python does not have a native array implementation, the `array` module provides basic array functionality.

from array import array

# Creating an array of integers
arr = array('i', [1, 2, 3, 4, 5])

# Accessing elements
print(arr[0])

# Appending an element
arr.append(6)
print(arr)

# Removing an element
arr.remove(3)
print(arr)

# Traversing the array
for num in arr:
    print(num, end=" ")
print("\n")

# Reversing the array
arr.reverse()
print(arr)

# Inserting an element at a specific position
arr.insert(2, 99)
print(arr)
```

## Lists
A **list** is a built-in Python data structure that is more versatile than arrays. It can store elements of different data types and dynamically resize as needed.

### Properties:
1. **Dynamic Size**: Lists can grow or shrink as elements are added or removed.
2. **Indexing**: Elements can be accessed using their index.
3. **Heterogeneous**: Can store elements of different types.

### Basic Operations:
- **Accessing Elements**: O(1) time complexity.
- **Inserting/Deleting Elements**: O(n) time complexity (due to shifting).
- **Traversal**: O(n) time complexity.

```python
# Creating a list
lst = [1, 2, 3, 4, 5]

# Accessing elements
print(lst[0])

# Appending an element
lst.append(6)
print(lst)

# Inserting at a specific position
lst.insert(2, 99)  # Insert 99 at index 2
print(lst) 

# Removing an element
lst.remove(4)
print(lst) 

# Traversing the list
for num in lst:
    print(num, end=" ")  
print("\n")

### Slicing: Extract sublists.
sublist = lst[1:4]
print(sublist)  


### List Comprehensions: Create lists using a concise syntax.
squared = [x**2 for x in range(5)]
print(squared)
```

### Comparing Arrays and Lists:
| Feature              | Array               | List                |
|----------------------|---------------------|---------------------|
| Fixed Size           | Yes                | No                  |
| Homogeneous Elements | Yes                | No                  |
| Built-in Functions   | Limited            | Extensive           |
| Performance          | Faster (for fixed) | Slower              |

### Practice Problem: Rotate an Array/List
**Problem**: Given an array or list, rotate it to the right by `k` steps.

```python
def rotate(arr, k):
    n = len(arr)
    k %= n  # Handle cases where k > n
    return arr[-k:] + arr[:-k]

arr = [1, 2, 3, 4, 5]
k = 2
rotated = rotate(arr, k)
print(rotated)
```
