# Hash Tables

## Introduction
Hash tables, also known as hash maps, are a data structure that maps keys to values for highly efficient lookups. They are implemented using an array and a hash function to compute an index into the array of buckets or slots, from which the desired value can be found.

### Key Features:
1. **Constant Time Complexity for Lookup and Insert:** Average case is O(1).
2. **Efficient for Sparse Data:** Allows fast access and is memory efficient when the dataset has large key-value ranges.
3. **Collision Handling:** Methods like chaining or open addressing resolve index conflicts.

```python
class HashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        # Simple hash function: Modulo operation
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        for pair in self.table[index]:
            if pair[0] == key:
                pair[1] = value  # Update value if key exists
                return
        self.table[index].append([key, value])

    def search(self, key):
        index = self.hash_function(key)
        for pair in self.table[index]:
            if pair[0] == key:
                return pair[1]
        return None  # Key not found

    def delete(self, key):
        index = self.hash_function(key)
        for i, pair in enumerate(self.table[index]):
            if pair[0] == key:
                del self.table[index][i]
                return True
        return False


ht = HashTable(10)
ht.insert("name", "Alice")
ht.insert("age", 25)
print(ht.search("name"))
ht.delete("name")
print(ht.search("name")) 
```

### Analysis:
1. **Hash Function:** Determines index for a key. The better the hash function, the fewer collisions occur.
2. **Collision Handling:**
   - **Chaining:** Use a list at each index to store multiple key-value pairs.
   - **Open Addressing:** Use probing to find the next empty slot when a collision occurs.

## Time Complexity
1. **Insert:** O(1) average, O(n) worst (due to collisions).
2. **Search:** O(1) average, O(n) worst (due to collisions).
3. **Delete:** O(1) average, O(n) worst (due to collisions).

## Applications of Hash Tables
1. **Caching:** Storing computed values for reuse.
2. **Databases:** Indexing for fast query resolution.
3. **Data Deduplication:** Quickly check for duplicates in a dataset.
4. **Symbol Table in Compilers:** Map variable/function names to memory locations.

## Problems

### Example 1: Word Frequency Counter
Given a text, count the frequency of each word.

```python
def word_frequency(text):
    words = text.split()
    freq_table = {}
    for word in words:
        freq_table[word] = freq_table.get(word, 0) + 1
    return freq_table

text = "hash tables are fast and efficient"
print(word_frequency(text))
```

### Example 2: Two-Sum Problem
Find two numbers in an array that add up to a given target.

```python
def two_sum(nums, target):
    hash_table = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in hash_table:
            return [hash_table[complement], i]
        hash_table[num] = i

nums = [2, 7, 11, 15]
target = 9
print(two_sum(nums, target))
```

### Example 3: Checking Anagrams
Check if two strings are anagrams.

```python
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False
    freq_table = {}
    for char in s1:
        freq_table[char] = freq_table.get(char, 0) + 1
    for char in s2:
        if char not in freq_table or freq_table[char] == 0:
            return False
        freq_table[char] -= 1
    return all(value == 0 for value in freq_table.values())

print(is_anagram("listen", "silent")) 
print(is_anagram("hello", "world"))
```

## Advanced Topics
1. **Rehashing:** Dynamically resize hash tables when load factor exceeds a threshold.
2. **Custom Hash Functions:** Use domain-specific knowledge to design efficient hash functions.
3. **Perfect Hashing:** Avoid collisions entirely with minimal space usage.

```python
class ResizableHashTable(HashTable):
    def __init__(self, size):
        super().__init__(size)
        self.count = 0

    def insert(self, key, value):
        super().insert(key, value)
        self.count += 1
        if self.count / self.size > 0.7:  # Load factor threshold
            self._rehash()

    def _rehash(self):
        old_table = self.table
        self.size *= 2
        self.table = [[] for _ in range(self.size)]
        self.count = 0
        for bucket in old_table:
            for key, value in bucket:
                self.insert(key, value)


ht = ResizableHashTable(5)
for i in range(10):
    ht.insert(f"key{i}", i)
print(ht.search("key5"))
```
