# Introduction to Heaps

## What is a Heap?
A **heap** is a special type of binary tree that satisfies the **heap property**:
1. **Max-Heap:** Every parent node is greater than or equal to its child nodes.
2. **Min-Heap:** Every parent node is less than or equal to its child nodes.

### Properties of Heaps:
- **Shape Property:** A heap is a complete binary tree, meaning all levels are fully filled except possibly the last level, which is filled from left to right.
- **Heap Property:** As mentioned above, the parent-child relationship depends on whether it is a max-heap or min-heap.

### Applications of Heaps:
1. **Priority Queues:** Heaps are often used to implement priority queues efficiently.
2. **Heap Sort:** A comparison-based sorting algorithm that uses a heap.
3. **Median Finding:** Used in dynamic data structures for finding medians efficiently.

## Implementation of Heaps in Python
Heaps are often implemented using arrays (or lists in Python), as their properties allow easy index-based manipulation.

### Index Relationships in Array Representation:
1. Parent of node at index `i`: `(i - 1) // 2`
2. Left child of node at index `i`: `2 * i + 1`
3. Right child of node at index `i`: `2 * i + 2`

```python
import heapq

# Create a min-heap
min_heap = []

# Insert elements into the heap
heapq.heappush(min_heap, 10)
heapq.heappush(min_heap, 5)
heapq.heappush(min_heap, 20)
heapq.heappush(min_heap, 1)

print("Min-Heap:", min_heap)  # Output: [1, 5, 20, 10]

# Remove the smallest element
smallest = heapq.heappop(min_heap)
print("Smallest Element:", smallest)
print("Min-Heap After Removal:", min_heap)
```

#### Time Complexity:
- **Insertion:** O(log n) per element (heapify up).
- **Deletion (extract min):** O(log n) per element (heapify down).

### Max-Heap Implementation
Python's `heapq` library only provides a min-heap. To implement a max-heap, we can invert the values.

```python
# Create a max-heap
max_heap = []

# Insert elements into the heap (invert values for max-heap)
heapq.heappush(max_heap, -10)
heapq.heappush(max_heap, -5)
heapq.heappush(max_heap, -20)
heapq.heappush(max_heap, -1)

print("Max-Heap:", [-x for x in max_heap]) 

# Remove the largest element
largest = -heapq.heappop(max_heap)
print("Largest Element:", largest)
print("Max-Heap After Removal:", [-x for x in max_heap]) 
```

## Problems and Solutions

### Problem 1: Find K Smallest/Largest Elements
Find the k smallest or largest elements from an unsorted array using a heap.

```python
def k_smallest_elements(arr, k):
    return heapq.nsmallest(k, arr)

def k_largest_elements(arr, k):
    return heapq.nlargest(k, arr)

arr = [7, 10, 4, 3, 20, 15]
k = 3

print("K Smallest Elements:", k_smallest_elements(arr, k)) 
print("K Largest Elements:", k_largest_elements(arr, k))  
```

#### Time Complexity:
- **O(n log k)** for both smallest and largest calculations.

### Problem 2: Heap Sort
Use a heap to sort an array in ascending or descending order.

```python
def heap_sort(arr):
    heapq.heapify(arr)  # Convert the list into a heap
    sorted_arr = [heapq.heappop(arr) for _ in range(len(arr))]
    return sorted_arr

arr = [7, 10, 4, 3, 20, 15]
print("Heap Sort Result:", heap_sort(arr))  # Output: [3, 4, 7, 10, 15, 20]
```

#### Time Complexity:
- **O(n log n)**: Building the heap is O(n), and extracting elements is O(log n) for each of the n elements.

### Problem 3: Merge K Sorted Arrays
Given `k` sorted arrays, merge them into one sorted array using a heap.

```python
def merge_k_sorted_arrays(arrays):
    min_heap = []

    # Insert the first element of each array along with its index
    for i, array in enumerate(arrays):
        if array:
            heapq.heappush(min_heap, (array[0], i, 0))

    result = []
    while min_heap:
        val, list_idx, elem_idx = heapq.heappop(min_heap)
        result.append(val)

        # Add the next element of the same array to the heap
        if elem_idx + 1 < len(arrays[list_idx]):
            heapq.heappush(min_heap, (arrays[list_idx][elem_idx + 1], list_idx, elem_idx + 1))

    return result


arrays = [[1, 5, 9], [2, 6], [3, 7, 8]]
print("Merged Array:", merge_k_sorted_arrays(arrays))
```

#### Time Complexity:
- **O(n log k):** Insertion/removal in the heap takes O(log k), and we process `n` elements in total.

## Advanced Topics

2. **Fibonacci Heaps:** More advanced heap structure with better amortized time for certain operations.
3. **Heapify Algorithm:** Building a heap in-place from an unsorted array in **O(n)** time.
