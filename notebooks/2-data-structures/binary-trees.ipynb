# Binary Trees

## Introduction
A binary tree is a hierarchical data structure in which each node has at most two children, referred to as the left child and the right child. Binary trees are fundamental in computer science and are used for various applications such as searching, sorting, and hierarchical storage.

---

## Properties of Binary Trees
1. **Maximum Nodes:** A binary tree of height \(h\) has at most \(2^h - 1\) nodes.
2. **Height of Tree:** The number of edges on the longest path from the root to a leaf.
3. **Full Binary Tree:** Every node has either 0 or 2 children.
4. **Complete Binary Tree:** All levels are fully filled except possibly the last, which is filled from left to right.
5. **Perfect Binary Tree:** All internal nodes have two children, and all leaves are at the same level.
6. **Balanced Binary Tree:** The difference between the heights of left and right subtrees of any node is not more than one.

---

## Representation of Binary Trees in Python
Binary trees can be represented using classes and nodes.

```python
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

class BinaryTree:
    def __init__(self):
        self.root = None

bt = BinaryTree()
bt.root = Node(1)
bt.root.left = Node(2)
bt.root.right = Node(3)
bt.root.left.left = Node(4)
bt.root.left.right = Node(5)
```

## Traversals in Binary Trees
Tree traversal refers to visiting each node in the tree systematically. There are three primary types of depth-first traversals:

1. **Inorder Traversal (Left, Root, Right):**
   - Visit the left subtree.
   - Visit the root node.
   - Visit the right subtree.

2. **Preorder Traversal (Root, Left, Right):**
   - Visit the root node.
   - Visit the left subtree.
   - Visit the right subtree.

3. **Postorder Traversal (Left, Right, Root):**
   - Visit the left subtree.
   - Visit the right subtree.
   - Visit the root node.

4. **Level-order Traversal (Breadth-First):**
   - Visit nodes level by level from top to bottom.



```python
def inorder(root):
    if root:
        inorder(root.left)
        print(root.key, end=" ")
        inorder(root.right)

def preorder(root):
    if root:
        print(root.key, end=" ")
        preorder(root.left)
        preorder(root.right)

def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.key, end=" ")

def level_order(root):
    if not root:
        return

    queue = [root]
    while queue:
        current = queue.pop(0)
        print(current.key, end=" ")

        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)

print("Inorder Traversal:")
inorder(bt.root)

print("\nPreorder Traversal:")
preorder(bt.root)

print("\nPostorder Traversal:")
postorder(bt.root)

print("\nLevel-order Traversal:")
level_order(bt.root)
```

### Problem 1: Height of a Binary Tree
Write a function to calculate the height of a binary tree.

```python
def height(root):
    if root is None:
        return 0
    return 1 + max(height(root.left), height(root.right))

print("Height of the tree:", height(bt.root))
```

#### Explanation:
- Recursively calculate the height of the left and right subtrees and return the maximum of the two plus one.
- Time Complexity: **O(n)** (Visit each node once).
- Space Complexity: **O(h)** (Call stack space, where h is the height).

### Problem 2: Check if a Binary Tree is Balanced
Write a function to check if a binary tree is height-balanced.

```python
def is_balanced(root):
    def check_balance(node):
        if not node:
            return 0, True

        left_height, left_balanced = check_balance(node.left)
        right_height, right_balanced = check_balance(node.right)

        balanced = left_balanced and right_balanced and abs(left_height - right_height) <= 1
        return 1 + max(left_height, right_height), balanced

    _, result = check_balance(root)
    return result


print("Is the tree balanced?:", is_balanced(bt.root))
```

#### Explanation:
- Recursively check the height of left and right subtrees and verify the balance condition.
- Time Complexity: **O(n)** (Visit each node once).
- Space Complexity: **O(h)**.

### Problem 3: Convert a Binary Tree to its Mirror
Write a function to convert a binary tree to its mirror image.

```python
def mirror(root):
    if root is None:
        return

    root.left, root.right = root.right, root.left
    mirror(root.left)
    mirror(root.right)

mirror(bt.root)
print("Inorder traversal of mirrored tree:")
inorder(bt.root)
```

#### Explanation:
- Swap the left and right children of each node recursively.
- Time Complexity: **O(n)**.
- Space Complexity: **O(h)**.
