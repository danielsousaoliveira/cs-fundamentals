# Linked Lists

Linked lists are a fundamental data structure that organizes data in a linear sequence, but unlike arrays, elements are not stored in contiguous memory locations. Instead, each element (node) contains a value and a reference (or pointer) to the next node in the sequence.

## Types of Linked Lists

1. **Singly Linked List**: Each node points to the next node.
2. **Doubly Linked List**: Each node points to both the next and the previous nodes.
3. **Circular Linked List**: The last node points back to the first node, creating a circular structure.

## Structure of a Node

Each node in a linked list typically has two components:
1. **Data**: Stores the value.
2. **Next**: Stores a reference (or pointer) to the next node.

For doubly linked lists, nodes also include:
3. **Prev**: Stores a reference to the previous node.

### Singly Linked List Implementation

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

    def delete_at_beginning(self):
        if not self.head:
            return
        self.head = self.head.next
    
    def delete_at_end(self):
        if not self.head:
            return
        if not self.head.next:
            self.head = None
            return
        current = self.head
        while current.next.next:
            current = current.next
        current.next = None

    def display(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
    
    def reverse(self):
        prev = None
        current = self.head
        
        while current:
            
            next_node = current.next # Store next node
            current.next = prev # Reverse the link
            prev = current # Move prev and current one step forward
            current = next_node
            
        self.head = prev # Update head

sll = SinglyLinkedList()
sll.insert_at_end(1)
sll.insert_at_end(2)
sll.insert_at_end(3)
sll.display() 
sll.reverse()
sll.display()
```

### Operations and Complexity:
1. **Insertion at Beginning**: O(1)
2. **Insertion at End**: O(n)
3. **Traversal/Display**: O(n)
4. **Search for a Value**: O(n)
5. **Deletion of a Node**:
   - From the Beginning: O(1)
   - From the End or Middle: O(n)

### Doubly Linked List Implementation

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
        new_node.prev = current

    def display_forward(self):
        current = self.head
        while current:
            print(current.data, end=" <-> ")
            current = current.next
        print("None")

    def display_backward(self):
        current = self.head
        while current and current.next:
            current = current.next
        while current:
            print(current.data, end=" <-> ")
            current = current.prev
        print("None")
    
    def delete_at_position(self, position):

        if not self.head:
            print("List is empty")
            return

        if position < 0:
            print("Invalid position")
            return

        current = self.head
        
        # Handle deletion at head (position 0)
        if position == 0:
            self.head = current.next
            if self.head:  # If list had more than one node
                self.head.prev = None
            return

        # Move to the position
        for _ in range(position):
            if current is None:
                print("Position out of range")
                return
            current = current.next

        # If position is out of range
        if current is None:
            print("Position out of range")
            return

        # Handle deletion at the end
        if current.next is None:
            current.prev.next = None
            return

        # Handle deletion in the middle
        current.prev.next = current.next
        current.next.prev = current.prev

dll = DoublyLinkedList()
dll.insert_at_end(1)
dll.insert_at_end(2)
dll.insert_at_end(3)
dll.display_forward()  
dll.display_backward()

dll.delete_at_position(1)  # Delete middle node (2)
print("After deleting position 1:")
dll.display_forward() 
dll.display_backward() 
```

### Operations and Complexity:
1. **Insertion at Beginning**: O(1)
2. **Insertion at End**: O(n)
3. **Traversal (Forward or Backward)**: O(n)
4. **Search for a Value**: O(n)
5. **Deletion of a Node**:
   - From the Beginning: O(1)
   - From the End or Middle: O(n)

## Circular Linked List

In a circular linked list, the last node’s `next` pointer links back to the head.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None

    def insert_at_end(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            new_node.next = self.head
            return
        current = self.head
        while current.next != self.head:
            current = current.next
        current.next = new_node
        new_node.next = self.head

    def display(self):
        if not self.head:
            return
        current = self.head
        while True:
            print(current.data, end=" -> ")
            current = current.next
            if current == self.head:
                break
        print("(head)")

def detect_cycle(lista):
    # Empty list or single node
    if not lista.head or not lista.head.next:
        return False
        
    slow = fast = lista.head
    
    while fast and fast.next:
        slow = slow.next         # move one step
        fast = fast.next.next    # move two steps
        
        if slow == fast:
            return True
            
    return False

cll = CircularLinkedList()
cll.insert_at_end(1)
cll.insert_at_end(2)
cll.insert_at_end(3)
cll.display()
print("Is circular:", detect_cycle(cll)) 

sll = SinglyLinkedList()
sll.insert_at_end(1)
sll.insert_at_end(2)
sll.insert_at_end(3)
sll.display()
print("Is circular:", detect_cycle(sll)) 
```

### Operations and Complexity:
1. **Insertion at Beginning**: O(1)
2. **Insertion at End**: O(n)
3. **Traversal**: O(n)
4. **Search for a Value**: O(n)
5. **Deletion of a Node**:
   - From the Beginning: O(1)
   - From the End or Middle: O(n)


## Applications of Linked Lists
1. **Dynamic Memory Allocation**: Useful for dynamic data where array resizing would be inefficient.
2. **Implementation of Data Structures**: Stacks, queues, and graphs.
3. **Undo Functionality**: Store operations in applications.
4. **Hash Tables**: To resolve collisions using chaining.
5. **Operating Systems**: Represent file directory structures.

## Comparison with Arrays
| Feature                | Linked List              | Array                     |
|------------------------|--------------------------|---------------------------|
| Memory Allocation      | Dynamic                 | Static or Dynamic         |
| Access Time            | O(n) (sequential)       | O(1) (random)             |
| Insertion/Deletion     | Efficient (O(1) or O(n))| Inefficient (O(n))        |
| Cache Locality         | Poor                    | Good                      |

## Benchmark insertion and deletion for linked lists and arrays

```python
import time

def benchmark_operation(operation, size, iterations):
    total_time = 0
    for _ in range(iterations):
        start_time = time.time()
        operation(size)
        total_time += time.time() - start_time
    return total_time / iterations

def benchmark_linked_list_operations(size):
    # Linked List operations
    def test_ll_insert_beginning():
        ll = SinglyLinkedList()
        for i in range(size):
            ll.insert_at_beginning(i)
    
    def test_ll_insert_end():
        ll = SinglyLinkedList()
        for i in range(size):
            ll.insert_at_end(i)
    
    def test_ll_delete_beginning():
        ll = SinglyLinkedList()
        for i in range(size):
            ll.insert_at_beginning(i)
        for _ in range(size):
            ll.delete_at_beginning()
    
    def test_ll_delete_end():
        ll = SinglyLinkedList()
        for i in range(size):
            ll.insert_at_beginning(i)
        for _ in range(size):
            ll.delete_at_end()
    
    return {
        "Insert at beginning": benchmark_operation(lambda x: test_ll_insert_beginning(), size, 5),
        "Insert at end": benchmark_operation(lambda x: test_ll_insert_end(), size, 5),
        "Delete from beginning": benchmark_operation(lambda x: test_ll_delete_beginning(), size, 5),
        "Delete from end": benchmark_operation(lambda x: test_ll_delete_end(), size, 5)
    }

def benchmark_array_operations(size):
    # Array (list in Python) operations
    def test_array_insert_beginning():
        arr = []
        for i in range(size):
            arr.insert(0, i)
    
    def test_array_insert_end():
        arr = []
        for i in range(size):
            arr.append(i)
    
    def test_array_delete_beginning():
        arr = list(range(size))
        for _ in range(size):
            arr.pop(0)
    
    def test_array_delete_end():
        arr = list(range(size))
        for _ in range(size):
            arr.pop()
    
    return {
        "Insert at beginning": benchmark_operation(lambda x: test_array_insert_beginning(), size, 5),
        "Insert at end": benchmark_operation(lambda x: test_array_insert_end(), size, 5),
        "Delete from beginning": benchmark_operation(lambda x: test_array_delete_beginning(), size, 5),
        "Delete from end": benchmark_operation(lambda x: test_array_delete_end(), size, 5)
    }

def run_benchmarks(sizes):
    print("Running benchmarks...\n")
    print(f"{'Operation':<25} {'Size':<10} {'Linked List':<15} {'Array':<15} {'Faster'}")
    print("-" * 70)
    
    for size in sizes:
        ll_results = benchmark_linked_list_operations(size)
        array_results = benchmark_array_operations(size)
        
        for operation in ll_results.keys():
            ll_time = ll_results[operation]
            array_time = array_results[operation]
            faster = "Linked List" if ll_time < array_time else "Array"
            
            print(f"{operation:<25} {size:<10} {ll_time:.6f}s {'':2} {array_time:.6f}s {'':2} {faster}")
        print("-" * 70)

sizes = [100, 1000, 10000]
run_benchmarks(sizes)
```
